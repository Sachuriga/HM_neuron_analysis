# hm_neuron quickstart
Load a session, look at its place fields, and run an analysis from Python.

```
from hm_neuron.config import Config
from hm_neuron import io, fields, maze, plotting
from hm_neuron import runner
cfg = Config()
cat = io.catalog(cfg)
cat.head()
```

```
s = io.load_session(cfg, 'Rat5', '20260709')
R = fields.session_fields(s, cfg)
R['units'].sort_values('si', ascending=False).head()
```

## Rate map of the most informative place cell on the maze

```
import numpy as np
lm = maze.maze_from_cfg(cfg)
k = int(np.nanargmax(np.where(R['units'].is_place, R['units'].si, np.nan)))
fig, ax = plotting.figure(120, 70)
plotting.draw_maze(ax, lm, R['rate'][k], cmap='magma', goal=s.goal, colorbar_label='Hz')
ax.set_title(f"unit {R['uids'][k]}  SI {R['units'].si.iloc[k]:.2f}")
fig
```

## Run an analysis (same as `hm run field_count`)

```
res = runner.run_analysis('field_count', cfg, rats=['Rat5'])
res['fits'].head()
```
